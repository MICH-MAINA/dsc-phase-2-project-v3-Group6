```python


# Define the query
query = """
SELECT
    movie_basics.movie_id,
    movie_basics.primary_title,
    movie_basics.original_title,
    movie_basics.start_year,
    movie_basics.runtime_minutes,
    movie_basics.genres,

    movie_ratings.averagerating,
    movie_ratings.numvotes,

    movie_akas.title AS aka_title,
    movie_akas.region,
    movie_akas."language",
    movie_akas."types",
    movie_akas."attributes",
    movie_akas.is_original_title,

    principals.ordering AS principal_order,
    principals.category,
    principals.job,
    principals.characters,

    persons.person_id,
    persons.primary_name,
    persons.birth_year,
    persons.death_year,
    persons.primary_profession,

    directors.person_id AS director_id,
    writers.person_id AS writer_id,
    known_for.person_id AS known_for_id

FROM movie_basics

-- Ratings
LEFT JOIN movie_ratings 
    ON movie_basics.movie_id = movie_ratings.movie_id

-- Alternative titles
LEFT JOIN movie_akas 
    ON movie_basics.movie_id = movie_akas.movie_id

-- Principals (cast/crew)
LEFT JOIN principals 
    ON movie_basics.movie_id = principals.movie_id

-- Person details
LEFT JOIN persons 
    ON principals.person_id = persons.person_id

-- Directors
LEFT JOIN directors 
    ON movie_basics.movie_id = directors.movie_id

-- Writers
LEFT JOIN writers 
    ON movie_basics.movie_id = writers.movie_id

-- Known-for (people known for this movie)
LEFT JOIN known_for 
    ON movie_basics.movie_id = known_for.movie_id;



"""

# Run the query and load into DataFrame
df = pd.read_sql(query, conn)

# Show first few rows
df.head()

# Close the connection (optional if you'll reuse it later)
#conn.close()
```
